# Benchmark Results Evaluation

This notebook demonstrates how to evaluate and visualize benchmark results using the snomed_methods benchmarking suite.

```python
# Set matplotlib backend for non-interactive use (no windows in VS Code)
import matplotlib

matplotlib.use("Agg")

import os
from pathlib import Path

from snomed_methods.benchmarking import (
    aggregate_results,
    evaluate_suite,
    load_results,
    save_results,
    summarize_benchmark,
)
```

## Define Benchmark Methods to Evaluate

This notebook is designed to work with a `benchmarks.yaml` configuration file that specifies which benchmark methods to run and compare.

```python
# Check for benchmarks.yaml configuration
config_path = "benchmarks.yaml"
if os.path.exists(config_path):
    import yaml

    with open(config_path) as f:
        config = yaml.safe_load(f)
    print("Configuration loaded from", config_path)
    print(f"Found {len(config.get('benchmarks', []))} benchmark methods")
else:
    print(f"No configuration file found at {config_path}")
```

## Overview: Summary Statistics

```python
# Generate summary statistics
summary = summarize_benchmark(benchmark_results)
print("Summary Statistics:")
summary
```

## Comparison: Aggregate Results

Convert benchmark results to a DataFrame for easy analysis:

```python
# Convert to DataFrame
results_df = aggregate_results(benchmark_results)
print("Benchmark Results DataFrame:")
results_df
```

## Evaluate Suite Results

Evaluate all benchmarks and save results with plots to an output directory:

```python
# Evaluate the benchmark suite (save to output directory)
output_dir = Path("benchmark_outputs")
results_df = evaluate_suite(benchmark_results, output_dir=output_dir)

print(f"Results saved to {output_dir}/")
print("Files in output directory:")
for f in sorted(output_dir.glob("*")):
    print(f"  - {f.name} ({f.stat().st_size} bytes)")
```

## Export Results

Save benchmark results in different formats for reporting:

```python
# Save to various formats
save_results(benchmark_results, "results.csv", file_format="csv")
save_results(benchmark_results, "results.json", file_format="json")

print("Results saved to:")
print("  - results.csv")
print("  - results.json")
```
