# Concept Mapping Demo

Demonstrates SNOMED CT concept mapping to ICD-10, LOINC, and RxNorm.

```python
from snomed_methods import ConceptMapper
```

```python
mapper = ConceptMapper()
print("ConceptMapper initialized")
```

```python
# Basic ICD-10 mapping
icd_mappings = mapper.map_to_icd("409681000000102")
print(f"ICD-10 mappings found: {len(icd_mappings)}")
for m in icd_mappings[:3]:
    print(
        f"{m['icd_code']} (conf: {m['confidence']:.2f}) - {m.get('description', 'N/A')}"
    )
```

```python
# Batch mapping
from snomed_methods import batch_map_concepts

results = batch_map_concepts(["409681000000102", "154621002"], ["ICD-10"])
print(f"Batch mapping results: {len(results)} concepts mapped")
for cui, mappings in results.items():
    for m in mappings.get("ICD-10", []):
        print(f"{cui} -> ICD {m['icd_code']} (conf: {m['confidence']:.2f})")
```
