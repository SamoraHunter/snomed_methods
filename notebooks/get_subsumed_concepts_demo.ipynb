# get_subsumed_concepts() Demo

This notebook demonstrates the `get_subsumed_concepts()` method for SNOMED CT hierarchy traversal.

```python
# Import the utility function that wraps all complexity
# Add project root to path for local module imports
import os
import sys

CURRENT_DIR = (
    os.path.dirname(os.path.abspath(__file__))
    if "__file__" in locals()
    else os.getcwd()
)
PROJECT_ROOT = os.path.abspath(os.path.join(CURRENT_DIR, ".."))
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

from src.snomed_methods.examples.get_subsumed_concepts_utils import (
    get_subsumed_concepts_for_term,
)
```

## Find Meningioma Subsumed Concepts

Traverse the is_a hierarchy to find all related concepts for 'meningioma'.

```python
# Get ancestors only
ids, names = get_subsumed_concepts_for_term(
    "meningioma", max_depth=10, include_ancestors=True, include_descendants=False
)

print(f"Found {len(ids)} ancestor concepts:")
for cui in ids:
    print(f"  - CUI: {cui}")
```

```python
# Get descendants only
ids, names = get_subsumed_concepts_for_term(
    "meningioma", max_depth=10, include_ancestors=False, include_descendants=True
)

print(f"Found {len(ids)} descendant concepts:")
for cui in ids:
    print(f"  - CUI: {cui}")
```

```python
# Get full DAG (both ancestors and descendants)
ids, names = get_subsumed_concepts_for_term(
    "meningioma", max_depth=10, include_ancestors=True, include_descendants=True
)

print(f"Found {len(ids)} concepts in full DAG:")
for cui in ids:
    print(f"  - CUI: {cui}")
```

## Multiple Terms

Process a list of terms and get all related concepts.

```python
# Import the multi-term function
from src.snomed_methods.examples.get_subsumed_concepts_utils import (
    get_subsumed_concepts_for_terms,
)
```

```python
# Process multiple terms
result = get_subsumed_concepts_for_terms(["meningioma"], max_depth=5)

print(f"Configuration used: {result['config']}")
print(f"Total unique concepts found: {len(result['all_concepts'])}")

for term, cui, concept_ids in result["results"]:
    print(f"\n'{term}' (CUI: {cui}):")
    print(f"  Found {len(concept_ids)} related concepts")
```
