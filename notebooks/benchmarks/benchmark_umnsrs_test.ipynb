# UMNSRS Benchmark Test

This notebook tests the benchmarking utilities with real UMNSRS dataset (requires HuggingFace access).

```python
# Import required modules
from snomed_methods.benchmarking.umnsrs import (
    download_umnsrs,
    evaluate_model,
    get_umnsrs_pairs,
)
```

## Download and Inspect Dataset

```python
# Download the relatedness subset
dataset = download_umnsrs(subset="relatedness")

print(f"Dataset size: {len(dataset)} concept pairs")
print(f"Columns: {dataset.column_names}")

# Preview data
df = dataset.to_pandas()
print("\nFirst 5 samples:")
for i in range(min(5, len(df))):
    print(
        f"  {i}: '{df.iloc[i]['text_1']}' vs '{df.iloc[i]['text_2']}' -> {df.iloc[i]['label']}",
    )
```

## Evaluate a Mock Method

```python
# Simple placeholder similarity function
def mock_similarity(text1: str, text2: str) -> float:
    """Mock method for testing - replace with your actual implementation."""
    # Example: length-based scoring
    return 500.0 + (len(text1) - len(text2)) * 10


# Get pairs and evaluate
pairs = get_umnsrs_pairs(subset="relatedness", cache_dir=None)

results = evaluate_model(
    model_func=mock_similarity,
    dataset_pairs=pairs[:50],
    metric=["spearman", "mae"],
)

print("\nBenchmark Results:")
for metric, value in results.items():
    print(f"  {metric}: {value:.4f}")
```
