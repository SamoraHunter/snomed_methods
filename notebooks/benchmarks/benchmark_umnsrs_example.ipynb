## Compare Different Subsets

```python
from snomed_methods.benchmarking.umnsrs import download_umnsrs

subsets = ["relatedness", "similarity"]

for subset in subsets:
    dataset = download_umnsrs(subset=subset)
    scores = dataset.to_pandas()["label"].tolist()

    print(f"\n{subset.upper()} ({len(dataset)} pairs):")
    print(f"  Score range: [{min(scores):.1f}, {max(scores):.1f}]")
    print(f"  Mean score: {sum(scores)/len(scores):.2f}")
```
